```python
def preprocess_text(text):

    doc = nlp(text)

    processed_words = []

    for token in doc:

        # Ignore punctuation
        if token.is_punct:
            continue

        # Ignore stopwords
        if token.is_stop:
            continue

        # Ignore spaces
        if token.is_space:
            continue

        # Lemmatize
        word = token.lemma_.lower()

        processed_words.append(word)

    return processed_words
```

```python
def find_keyword_matches(text, keyword):

    pattern = r'\b' + re.escape(keyword.lower()) + r'\b'

    matches = re.findall(
        pattern,
        text.lower()
    )

    return len(matches)
```

```python
print(find_keyword_matches(text, "students"))
print(find_keyword_matches(text, "EWS"))
print(find_keyword_matches(text, "women"))
```

```python
def analyze_stakeholders(text, stakeholder_categories):

    results = []

    for category, keywords in stakeholder_categories.items():

        matched_terms = []
        total_mentions = 0

        for keyword in keywords:

            count = find_keyword_matches(
                text,
                keyword
            )

            if count > 0:
                matched_terms.append(keyword)
                total_mentions += count

        if total_mentions > 0:
            status = "Detected"
        else:
            status = "Not detected"

        results.append({
            "Stakeholder": category,
            "Status": status,
            "Total Mentions": total_mentions,
            "Matched Terms": ", ".join(matched_terms)
        })

    return pd.DataFrame(results)
```

```python
# Split the document into sections based on 'CHAPTER' or 'SECTION' patterns
# Adjust this regex based on the actual document structure for more accurate sectioning
section_pattern = r'CHAPTER\s+[IVXLCDM]+\s*[\r\n]|SECTION\s+\d+\s*[\r\n]'

# Split the text by the pattern, keeping the delimiters to identify sections
raw_sections = re.split(section_pattern, text)

sections = {}
current_section_name = "Introduction/Preamble"

# Reconstruct sections with names. The re.split behavior is that it returns the split parts
# and the delimiters themselves. We need to pair them up.
for i, part in enumerate(raw_sections):
    if i == 0: # First part is usually introduction
        sections[current_section_name] = part.strip()
    else:
        # The split pattern will yield a delimiter, then the content following it
        # So we need to correctly assign the name from the split pattern if it exists
        match = re.search(section_pattern, text)
        if match and match.group(0):
            # Extract the actual section name from the match or generate one
            current_section_name = match.group(0).strip()
        else:
            current_section_name = f"Section {i}"
        sections[current_section_name] = part.strip()

# A more robust way would be to iterate through matches and content together
# For simplicity, for now, let's assume raw_sections alternates name and content for the most part
# Or, given this is an explanation of an error and subsequent fix, a direct fix is better.

# Let's refine the sectioning for better accuracy based on the observed text structure.
# Re-doing the section parsing for robustness

sections = {}
current_section_title = "Document Start"
current_section_content = []

lines = text.splitlines()
for line in lines:
    chapter_match = re.match(r'^(CHAPTER\s+[IVXLCDM]+)', line, re.IGNORECASE)
    section_match = re.match(r'^(\d+\.\s*[A-Z][a-zA-Z\s]+)', line)

    if chapter_match:
        if current_section_content:
            sections[current_section_title] = "\n".join(current_section_content).strip()
        current_section_title = chapter_match.group(1).strip()
        current_section_content = [line]
    elif section_match and not current_section_title.startswith("CHAPTER"): # Treat top-level sections if not already in a chapter
        if current_section_content and current_section_title != "Document Start":
             sections[current_section_title] = "\n".join(current_section_content).strip()
        current_section_title = section_match.group(1).strip()
        current_section_content = [line]
    else:
        current_section_content.append(line)

# Add the last accumulated section
if current_section_content:
    sections[current_section_title] = "\n".join(current_section_content).strip()


section_results = []

for section_name, section_text in sections.items():
    # Ensure section_text is not empty to avoid errors with textstat
    if not section_text.strip():
        continue

    score = textstat.flesch_reading_ease(
        section_text
    )

    word_count = len(
        re.findall(
            r'\b\w+\b',
            section_text
        )
    )

    section_results.append({

        "Section": section_name,

        "Word Count": word_count,

        "Readability Score": round(
            score,
            2
        )
    })

section_df = pd.DataFrame(
    section_results
)

section_df
```

```python
def find_sentences_with_keywords(text, keywords):

    results = []

    sentences = re.split(
        r'[.!?]+',
        text
    )

    for sentence in sentences:

        sentence_lower = sentence.lower()

        for keyword in keywords:

            if keyword.lower() in sentence_lower:

                results.append({
                    "Keyword": keyword,
                    "Sentence": sentence.strip()
                })

                break

    return pd.DataFrame(results)
```

```python
def find_stakeholder_restriction_sentences(
    text,
    stakeholder_words,
    restriction_words
):

    results = []

    sentences = re.split(
        r'[.!?]+',
        text
    )

    for sentence in sentences:

        sentence_lower = sentence.lower()

        found_stakeholders = [
            word
            for word in stakeholder_words
            if word.lower() in sentence_lower
        ]

        found_restrictions = [
            word
            for word in restriction_words
            if word.lower() in sentence_lower
        ]

        if found_stakeholders and found_restrictions:

            results.append({

                "Stakeholders":
                    ", ".join(found_stakeholders),

                "Restriction Terms":
                    ", ".join(found_restrictions),

                "Sentence":
                    sentence.strip()
            })

    return pd.DataFrame(results)
```
